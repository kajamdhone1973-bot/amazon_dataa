```python

import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# 1. Import CSV file
df = pd.read_csv("amazon dataa.csv")

# 2. Remove extra spaces from column names
df.columns = df.columns.str.strip()

# 3. Replace spaces and special characters in column names
df.columns = (
    df.columns
    .str.replace(" ", "_", regex=False)
    .str.replace(r"[^\w]", "", regex=True)
)

# 4. Check cleaned column names
print(df.columns.tolist())

# 5. Connect to MySQL
password = quote_plus("401620Nihal.")

engine = create_engine(
    f"mysql+pymysql://root:{password}@localhost:3306/amazon_sales"
)

# 6. Import data into MySQL
df.to_sql(
    name="amazon_data",
    con=engine,
    if_exists="replace",
    index=False
)

print("Data imported successfully!")
```
